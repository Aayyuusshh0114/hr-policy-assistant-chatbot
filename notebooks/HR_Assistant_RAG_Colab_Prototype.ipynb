```python
HR_GUARDRAILS = """
You are an HR Policy Assistant.

Rules:
1. Answer only HR-related questions.
2. Use only the information provided in the retrieved HR policy context.
3. Do not invent policies, benefits, rules, dates, amounts, or procedures.
4. If the answer is not supported by the context, respond:
   "I could not find this information in the available HR policies."
5. Do not treat general knowledge as company policy.
6. Keep answers clear, concise, and professional.
"""
```

```python
def is_hr_question(question):
    hr_keywords = [
        "leave", "payroll", "salary", "employee", "referral",
        "promotion", "insurance", "shift", "allowance",
        "performance", "separation", "notice", "job",
        "policy", "benefits", "dress code"
    ]

    question = question.lower()

    return any(keyword in question for keyword in hr_keywords)
    print(is_hr_question("What is the leave policy?"))
print(is_hr_question("Who won the cricket match yesterday?"))
```

```python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_template("""
{guardrails}

Use the following HR policy context to answer the question.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
def ask_hr_assistant(question):

    # 1. Scope guardrail
    if not is_hr_question(question):
        return {
            "answer": "I can only help with HR-related questions.",
            "sources": []
        }

    # 2. Retrieve relevant documents
    retrieved_docs = retriever.invoke(question)

    # 3. Build context
    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    # 4. Build prompt
    messages = rag_prompt.invoke({
        "guardrails": HR_GUARDRAILS,
        "context": context,
        "question": question
    })

    # 5. Generate answer
    response = llm.invoke(messages)

    # 6. Provenance
    sources = []

    for doc in retrieved_docs:
        source = doc.metadata.get("source", "Unknown source")
        page = doc.metadata.get("page", 0) + 1
        sources.append(f"{source} - Page {page}")

    sources = list(dict.fromkeys(sources))

    return {
        "answer": response.content,
        "sources": sources
    }
```

```python
def evaluate_answer(question, answer, context):

    evaluation_prompt = f"""
You are evaluating an HR RAG system.

Question:
{question}

Retrieved Context:
{context}

Generated Answer:
{answer}

Evaluate the answer on two criteria from 1 to 5:

Faithfulness:
5 = completely supported by the context
1 = contains unsupported or invented information

Answer Relevance:
5 = directly and completely answers the question
1 = does not answer the question

Return ONLY in this format:

Faithfulness: <score>
Answer Relevance: <score>
"""

    evaluation = llm.invoke(evaluation_prompt)

    return evaluation.content
```

```python
chat_history = []
chat_history.append({
    "user": "What is the leave policy?",
    "assistant": "Example previous answer"
})

print(chat_history)
```

```python
def ask_hr_assistant(question):

    # 1. Use recent conversation history
    recent_history = chat_history[-3:]

    history_text = "\n".join(
        f"User: {item['user']}\nAssistant: {item['assistant']}"
        for item in recent_history
    )

    # 2. Create standalone question for retrieval
    # This prevents raw conversation history from polluting retrieval.
    standalone_question = contextualize_question(
        question,
        chat_history
    )

    # 3. Retrieve relevant HR documents
    retrieved_docs = retriever.invoke(standalone_question)

    # Combine retrieved chunks into context
    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    # 4. Generate grounded answer
    prompt = f"""
{HR_GUARDRAILS}

You are an HR policy assistant.

Use ONLY the HR Policy Context below to answer the question.

Rules:
- Do not invent company policies.
- Do not use outside knowledge.
- If the provided context does not contain enough information
  to answer the question, say:
  "I could not find this information in the available HR policies."
- Answer clearly and concisely.

Previous conversation:
{history_text}

HR Policy Context:
{context}

Current Question:
{question}

Answer:
"""

    response = llm.invoke(prompt)
    answer = response.content.strip()

    # 5. Provenance / citations
    sources = []

    for doc in retrieved_docs:
        source = doc.metadata.get("source", "Unknown source")
        page = doc.metadata.get("page", 0) + 1
        sources.append(f"{source} - Page {page}")

    # Remove duplicate citations
    sources = list(dict.fromkeys(sources))

    # 6. If assistant refuses, don't display irrelevant sources
    refusal_phrases = [
        "could not find",
        "cannot find",
        "not available in the provided",
        "not found in the available",
        "does not contain"
    ]

    is_refusal = any(
        phrase in answer.lower()
        for phrase in refusal_phrases
    )

    if is_refusal:
        sources = []

    # 7. Store real conversation
    chat_history.append({
        "user": question,
        "assistant": answer
    })

    # 8. Return answer + verified retrieved sources
    return {
        "answer": answer,
        "sources": sources
    }
```

```python
audit_cases = [
    # Direct
    {
        "question": "What is the employee referral policy?",
        "expected_sources": ["Employee Referral Policy"],
        "tier": "direct"
    },
    {
        "question": "What is the leave management policy?",
        "expected_sources": ["Leave Management Policy"],
        "tier": "direct"
    },
    {
        "question": "What is the payroll policy?",
        "expected_sources": ["Payroll Policy"],
        "tier": "direct"
    },

    # Paraphrase
    {
        "question": "How can I refer my friend for a job here?",
        "expected_sources": ["Employee Referral Policy"],
        "tier": "paraphrase"
    },
    {
        "question": "What kinds of leave can I take?",
        "expected_sources": ["Leave Management Policy"],
        "tier": "paraphrase"
    },
    {
        "question": "When will my salary be processed?",
        "expected_sources": ["Payroll Policy"],
        "tier": "paraphrase"
    },

    # Scenario
    {
        "question": "I got another job and want to resign. What should I do?",
        "expected_sources": ["Separation Policy"],
        "tier": "scenario"
    },
    {
        "question": "I'm leaving the company next month. What's the process?",
        "expected_sources": ["Separation Policy"],
        "tier": "scenario"
    }
]
```

```python
def run_retrieval_audit(vector_store, audit_cases, k=5):

    results = []

    for case in audit_cases:

        question = case["question"]
        expected = case["expected_sources"]
        tier = case["tier"]

        docs = vector_store.similarity_search(question, k=k)

        sources = [
            doc.metadata.get("source", "UNKNOWN")
            for doc in docs
        ]

        def source_match(source):
            return any(
                expected_name.lower() in source.lower()
                for expected_name in expected
            )

        hit1 = any(source_match(src) for src in sources[:1])
        hit3 = any(source_match(src) for src in sources[:3])
        hit5 = any(source_match(src) for src in sources[:5])

        results.append({
            "question": question,
            "tier": tier,
            "sources": sources,
            "hit@1": hit1,
            "hit@3": hit3,
            "hit@5": hit5
        })

        print("\n" + "=" * 70)
        print("Tier:", tier)
        print("Question:", question)

        print("\nTop sources:")
        for i, source in enumerate(sources, 1):
            print(f"{i}. {source}")

        print(
            f"\nHit@1: {hit1} | "
            f"Hit@3: {hit3} | "
            f"Hit@5: {hit5}"
        )

    return results
```

```python
audit_results = run_retrieval_audit(
    vector_store,
    audit_cases,
    k=5
)
```

```python
from collections import defaultdict

tier_scores = defaultdict(
    lambda: {
        "count": 0,
        "hit1": 0,
        "hit3": 0,
        "hit5": 0
    }
)

for result in audit_results:

    tier = result["tier"]

    tier_scores[tier]["count"] += 1
    tier_scores[tier]["hit1"] += int(result["hit@1"])
    tier_scores[tier]["hit3"] += int(result["hit@3"])
    tier_scores[tier]["hit5"] += int(result["hit@5"])
```

```python
for tier, scores in tier_scores.items():

    total = scores["count"]

    print(f"\n{tier.upper()}")
    print("Hit@1:", round(scores["hit1"] / total, 2))
    print("Hit@3:", round(scores["hit3"] / total, 2))
    print("Hit@5:", round(scores["hit5"] / total, 2))
```

```python
def audit_chunk_quality(chunks):

    print("TOTAL CHUNKS:", len(chunks))

    lengths = [len(chunk.page_content) for chunk in chunks]

    print("Average length:", round(sum(lengths) / len(lengths)))
    print("Minimum length:", min(lengths))
    print("Maximum length:", max(lengths))

    short_chunks = [
        chunk for chunk in chunks
        if len(chunk.page_content.strip()) < 200
    ]

    print("Chunks under 200 characters:", len(short_chunks))
    print(
        "Short chunk percentage:",
        round(len(short_chunks) / len(chunks) * 100, 2),
        "%"
    )


audit_chunk_quality(chunks)
```

```python
import numpy as np

def cosine_similarity(a, b):
    a = np.array(a)
    b = np.array(b)

    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )


test_pairs = [
    ("resign", "resignation"),
    ("resign", "voluntary separation"),
    ("resign", "employee leaving"),

    ("notice period", "exit procedure"),
    ("notice period", "relieving"),

    ("refer my friend", "employee referral"),
    ("salary", "payroll"),

    ("intern", "full-time employee"),
    ("intern", "internal job posting"),

    ("harassment", "grievance"),
    ("harassment", "misconduct"),
]


for a, b in test_pairs:

    vectors = embeddings.embed_documents([a, b])

    score = cosine_similarity(
        vectors[0],
        vectors[1]
    )

    print(f"{a:20} <-> {b:25} = {score:.3f}")
```

```python
def contextualize_question(question, history):

    if not history:
        return question

    recent_history = history[-3:]

    history_text = "\n".join(
        f"User: {item['user']}\nAssistant: {item['assistant']}"
        for item in recent_history
    )

    prompt = f"""
You rewrite conversational HR questions into standalone search questions.

Use the conversation only to resolve references such as:
"it", "that", "they", "this policy", "what about that".

Rules:
- Preserve the user's intent.
- Do not answer the question.
- Do not invent information.
- If the question already makes sense by itself, return it unchanged.
- Return only the standalone question.

Conversation:
{history_text}

Current question:
{question}

Standalone question:
"""

    response = llm.invoke(prompt)

    return response.content.strip()
```
